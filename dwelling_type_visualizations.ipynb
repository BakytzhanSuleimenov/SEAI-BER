# BER Data Dwelling Type Visualizations
Comprehensive visualizations for Detached, Attached, and Apartment dwellings aggregated by County

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

sns.set_style('whitegrid')
sns.set_palette('husl')
plt.rcParams['figure.figsize'] = (16, 12)
```

```python
file_path = 'BERPublicsearch.txt'
print('Loading data...')
df = pd.read_csv(file_path, sep='\t', encoding='latin-1', engine='python', on_bad_lines='skip')
print(f'Data loaded: {df.shape}')
```

```python
detached = ['Detached house']
attached = ['Semi-detached house', 'Terraced house', 'End of terrace house', 'Mid-terrace house']
apartment = ['Apartment', 'Mid-floor apartment', 'Top-floor apartment', 'Ground-floor apartment', 'Maisonette']

detached_df = df[df['DwellingTypeDescr'].isin(detached)].copy()
attached_df = df[df['DwellingTypeDescr'].isin(attached)].copy()
apartment_df = df[df['DwellingTypeDescr'].isin(apartment)].copy()

print("Dwelling Type Distribution:")
print(f"Detached: {len(detached_df):,} records")
print(f"Attached: {len(attached_df):,} records")
print(f"Apartment: {len(apartment_df):,} records")
```

## DETACHED HOUSES - Aggregated by County

```python
dwelling_data = detached_df.copy()
top_counties = dwelling_data['CountyName'].value_counts().head(15).index
county_filtered = dwelling_data[dwelling_data['CountyName'].isin(top_counties)]

fig = plt.figure(figsize=(20, 24))
fig.suptitle('DETACHED HOUSES - Analysis by County (Top 15 Counties)', fontsize=18, fontweight='bold', y=0.995)

ax = plt.subplot(4, 2, 1)
year_by_county = county_filtered.groupby('CountyName')['Year_of_Construction'].mean().sort_values()
year_by_county.plot(kind='barh', ax=ax, color='steelblue')
ax.set_xlabel('Average Year of Construction')
ax.set_title('Year of Construction by County', fontweight='bold', fontsize=12)
ax.grid(axis='x', alpha=0.3)

ax = plt.subplot(4, 2, 2)
ber_by_county = county_filtered.groupby('CountyName')['BerRating'].mean().sort_values()
ber_by_county.plot(kind='barh', ax=ax, color='coral')
ax.set_xlabel('Average BER Rating')
ax.set_title('BER Rating by County', fontweight='bold', fontsize=12)
ax.grid(axis='x', alpha=0.3)

ax = plt.subplot(4, 2, 3)
area_by_county = county_filtered.groupby('CountyName')['GroundFloorArea(sq m)'].mean().sort_values()
area_by_county.plot(kind='barh', ax=ax, color='mediumseagreen')
ax.set_xlabel('Average Ground Floor Area (sq m)')
ax.set_title('Ground Floor Area by County', fontweight='bold', fontsize=12)
ax.grid(axis='x', alpha=0.3)

ax = plt.subplot(4, 2, 4)
rating_counts = county_filtered.groupby('TypeofRating')['CountyName'].count()
colors = plt.cm.Set3(range(len(rating_counts)))
ax.pie(rating_counts.values, labels=rating_counts.index, autopct='%1.1f%%', colors=colors, startangle=90)
ax.set_title('Type of Rating Distribution', fontweight='bold', fontsize=12)

ax = plt.subplot(4, 2, 5)
energy_rating_top = county_filtered['EnergyRating'].value_counts().head(12).sort_values()
energy_rating_top.plot(kind='barh', ax=ax, color='mediumpurple')
ax.set_xlabel('Count')
ax.set_title('Top 12 Energy Ratings', fontweight='bold', fontsize=12)
ax.grid(axis='x', alpha=0.3)

ax = plt.subplot(4, 2, 6)
heating_fuel = county_filtered['MainSpaceHeatingFuel'].value_counts().head(10)
colors = plt.cm.Set2(range(len(heating_fuel)))
ax.barh(range(len(heating_fuel)), heating_fuel.values, color=colors)
ax.set_yticks(range(len(heating_fuel)))
ax.set_yticklabels(heating_fuel.index, fontsize=9)
ax.set_xlabel('Count')
ax.set_title('Main Space Heating Fuel Types', fontweight='bold', fontsize=12)
ax.grid(axis='x', alpha=0.3)

ax = plt.subplot(4, 2, 7)
water_fuel = county_filtered['MainWaterHeatingFuel'].value_counts().head(10)
colors = plt.cm.Pastel1(range(len(water_fuel)))
ax.barh(range(len(water_fuel)), water_fuel.values, color=colors)
ax.set_yticks(range(len(water_fuel)))
ax.set_yticklabels(water_fuel.index, fontsize=9)
ax.set_xlabel('Count')
ax.set_title('Main Water Heating Fuel Types', fontweight='bold', fontsize=12)
ax.grid(axis='x', alpha=0.3)

ax = plt.subplot(4, 2, 8)
energy_by_county = county_filtered.groupby('CountyName')['EnergyRating'].apply(
    lambda x: x.value_counts().index[0] if len(x) > 0 else 'Unknown'
).value_counts()
colors = plt.cm.Spectral(np.linspace(0, 1, len(energy_by_county)))
ax.bar(range(len(energy_by_county)), energy_by_county.values, color=colors)
ax.set_xticks(range(len(energy_by_county)))
ax.set_xticklabels(energy_by_county.index, fontsize=10, fontweight='bold')
ax.set_ylabel('Frequency')
ax.set_title('Most Common Energy Rating by County', fontweight='bold', fontsize=12)
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('dwelling_detached_visualizations.png', dpi=300, bbox_inches='tight')
plt.show()

print("\nDetached Houses Summary Statistics:")
print(f"  Records: {len(county_filtered):,}")
print(f"  Avg Year of Construction: {county_filtered['Year_of_Construction'].mean():.1f}")
print(f"  Avg BER Rating: {county_filtered['BerRating'].mean():.2f}")
print(f"  Avg Ground Floor Area: {county_filtered['GroundFloorArea(sq m)'].mean():.2f} sq m")
```

## Comparative Summary

```python
summary_data = {
    'Dwelling Type': ['Detached', 'Attached', 'Apartment'],
    'Records': [len(detached_df), len(attached_df), len(apartment_df)],
    'Avg Year': [
        detached_df['Year_of_Construction'].mean(),
        attached_df['Year_of_Construction'].mean(),
        apartment_df['Year_of_Construction'].mean()
    ],
    'Avg BER Rating': [
        detached_df['BerRating'].mean(),
        attached_df['BerRating'].mean(),
        apartment_df['BerRating'].mean()
    ],
    'Avg Ground Floor Area (sq m)': [
        detached_df['GroundFloorArea(sq m)'].mean(),
        attached_df['GroundFloorArea(sq m)'].mean(),
        apartment_df['GroundFloorArea(sq m)'].mean()
    ]
}

summary_df = pd.DataFrame(summary_data)
print("\nComparative Summary of Dwelling Types:")
print(summary_df.to_string(index=False))

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Comparative Analysis: Detached vs Attached vs Apartment', fontsize=16, fontweight='bold')

axes[0, 0].bar(summary_df['Dwelling Type'], summary_df['Avg Year'], color=['steelblue', 'coral', 'mediumseagreen'])
axes[0, 0].set_ylabel('Average Year')
axes[0, 0].set_title('Average Year of Construction')
axes[0, 0].grid(axis='y', alpha=0.3)

axes[0, 1].bar(summary_df['Dwelling Type'], summary_df['Avg BER Rating'], color=['steelblue', 'coral', 'mediumseagreen'])
axes[0, 1].set_ylabel('Average BER Rating')
axes[0, 1].set_title('Average BER Rating')
axes[0, 1].grid(axis='y', alpha=0.3)

axes[1, 0].bar(summary_df['Dwelling Type'], summary_df['Avg Ground Floor Area (sq m)'], color=['steelblue', 'coral', 'mediumseagreen'])
axes[1, 0].set_ylabel('Average Area (sq m)')
axes[1, 0].set_title('Average Ground Floor Area')
axes[1, 0].grid(axis='y', alpha=0.3)

axes[1, 1].bar(summary_df['Dwelling Type'], summary_df['Records']/1000, color=['steelblue', 'coral', 'mediumseagreen'])
axes[1, 1].set_ylabel('Records (thousands)')
axes[1, 1].set_title('Total Records')
axes[1, 1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('dwelling_comparative_summary.png', dpi=300, bbox_inches='tight')
plt.show()

print("\nAll visualizations completed successfully!")
```
